Prompting

## Task Overview
- **Target Task:** Automated Code Review & Refactoring of Node.js/Express API route for Quranic Tajweed Learning App.
- **Objective:** Transform a vague code review request into an engineered, production-ready system prompt returning structured JSON.

---

## Prompt Iterations & Comparative Analysis

### Iteration 1: The Naive Prompt
- **Prompt:** `"This is my Quranic Tajweed Backend API, evaluate its accuracy and structure."`
- **Technique:** Baseline / Naive
- **Output:** High-level advice lacking code fixes or actionable security checks.

### Iteration 2: Role Assignment
- **Prompt:** `"You are a Senior Backend Engineer specializing in Node.js, Express, and RESTful API architecture. Review the following Quranic Tajweed API route, critique its security and structure, and provide a refactored version."`
- **Technique:** Role Assignment / Persona
- **Comparison vs V1:** Output quality improved dramatically by anchoring feedback to professional backend engineering standards rather than general advice.

### Iteration 3: Context & Motivation
- **Prompt:** `"You are a Senior Backend Engineer. Review this Node.js/Express API route for a Quranic Tajweed platform serving Tajweed audio lectures and articulation point metadata under high traffic. Unhandled rejections crash our server. Critique the route for memory connection leaks, async exception safety, and REST standards."`
- **Technique:** Operational Context & Constraints
- **Comparison vs V2:** Adding failure consequences shifted focus from code formatting to critical concurrency and server stability risks.

### Iteration 4: Few-Shot Examples
- **Prompt:** `"You are a Senior Backend Engineer reviewing a Node.js/Express Tajweed API endpoint. Example Input: app.get('/lectures/:id', async (req, res) => { const data = await db.query('SELECT * FROM lectures WHERE id = ' + req.params.id); res.send(data); }); Expected Output Structure: Vulnerability: SQL Injection. Refactored Code: Use parameterized queries. Evaluate this Tajweed lecture route using the same structure..."`
- **Technique:** Few-Shot In-Context Learning
- **Comparison vs V3:** Eliminated conversational fluff and standardized the audit format directly.

### Iteration 5: Output Structuring
- **Prompt:** `"...Structure output strictly in two parts: 1) Markdown summary table | Issue | Severity | Fix | and 2) Valid JSON object with audit_status and refactored_code."`
- **Technique:** Enforced Schema / Output Structuring
- **Comparison vs V4:** Converted free-form text into a machine-readable JSON object suitable for automated CI/CD pipelines.

### Iteration 6: Step Decomposition (Chain-of-Thought)
- **Prompt:** `"You are a Senior Backend Engineer. Before generating output, analyze step-by-step inside <analysis> tags: 1) SQL injection 2) Async error handling 3) HTTP status codes. Then output Markdown table and JSON refactor object."`
- **Technique:** Step Decomposition / Chain-of-Thought
- **Comparison vs V5:** Forcing step-by-step reasoning caught hidden async edge cases that V5 missed.
