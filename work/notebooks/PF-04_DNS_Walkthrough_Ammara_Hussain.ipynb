#**DNS Walkthrough**

**What is DNS?**

DNS (Domain Name System) translates human-readable domain names into the information needed to reach a website. Instead of remembering an IP address, users can type a domain such as example.com, and DNS helps their device find the correct server.

**What is a CNAME record?**

A CNAME (Canonical Name) record is a DNS record that makes one domain name an alias for another domain name. For example, when my FlyRank subdomain is provided in the future, a CNAME can be used to point my subdomain toward the hostname provided by my hosting platform, such as Netlify. The exact target value will depend on the configuration provided by FlyRank/Netlify.

**What happens when someone enters my website address?**

When someone types my domain into a browser, the browser first needs to determine where that domain is hosted. The request goes through a DNS resolver, which looks for the appropriate DNS information. The resolver contacts the domain's authoritative nameserver if the information is not already cached. The nameserver returns the relevant DNS record, such as a CNAME or A record.

If a CNAME is used, the DNS system follows that alias to the target hostname and obtains the information needed to reach the hosting service. The browser can then connect to the hosting provider, such as Netlify, and request my website files.

Finally, the website is delivered to the browser over HTTPS. HTTPS encrypts the connection between the visitor and the website and is represented by the padlock shown in the browser.
