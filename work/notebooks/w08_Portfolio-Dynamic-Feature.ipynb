# **How My Contact Form Works**
A backend is the part of a website or application that works behind the scenes to receive and process data. My portfolio was originally a static website, but for this assignment I added one dynamic feature: a contact form.

The contact form allows a visitor to enter their name, email address, and message. When the visitor clicks the "Send Message" button, the browser sends the form data to Netlify using an HTTP POST request.

Netlify handles the form submission for me, so I did not need to create my own server or database. Netlify receives the submitted information and makes it available in the Forms section of my Netlify dashboard.

In simple terms, the data flows from the visitor → the HTML form → the browser → Netlify → my Netlify Forms dashboard.

I tested the feature with a real submission and confirmed that the submission was successfully received by Netlify. This helped me understand how a static website can use a hosted service to provide a real dynamic feature without building a complete backend myself.
