Stack Selection Rationale: Three Roads Analysis
1. Context & Constraints Provided
Cost Constraint: 100% Free tier hosting and tools (no paid subscriptions/servers).

Skill Level: Undergraduate Software Engineering student (proficient in HTML/CSS, JavaScript, Node.js/Express, Python, and Oracle SQL).

Portfolio Goal: Display machine learning assignment outputs (dataframes/plots), backend REST API architectures (JSON outputs, endpoints), and software documentation.

Dynamic Needs: Not yet. Static hosting with clean presentation is sufficient for now; dynamic backend API features can be added later if needed.

2. The Three Stack Options Evaluated
Option 1: Simplest — Static HTML5 / CSS3 / Vanilla JS
How to Build: Pure semantic HTML5 markup styled with CSS3 and minimal JavaScript for theme/interactivity.

Where to Host: GitHub Pages (Free).

Backend Needed? No.

Trade-Off: Fast to deploy and complete zero build step, but requires manual copy-pasting for repetitive elements (like navigation bars or headers across pages).

Option 2: Middle / Recommended — Lightweight Static Framework (Astro or Eleventy)
How to Build: Use Markdown files for content/case studies compiled into lightweight HTML via component templates.

Where to Host: Vercel or GitHub Pages (Free).

Backend Needed? No.

Trade-Off: Modular components make updating content effortless, but adds a minor build step / dependency setup upfront.

Option 3: Most Powerful — React / Next.js Framework
How to Build: Full component-driven single-page application built with React and Tailwind CSS.

Where to Host: Vercel (Free).

Backend Needed? Optional (Serverless API routes available if needed later).

Trade-Off: Highly extensible and ideal for interactive web apps, but introduces unnecessary JavaScript bundle overhead, state management complexity, and longer setup time for a static portfolio.

3. Pressure-Testing the Front-Runner
What breaks if I pick the simplest (Option 1)? Nothing breaks. However, managing repetitive code across multiple case study pages becomes tedious without template components.

What do I maintain if I pick the most powerful (Option 3)? Node module dependencies, framework version updates, build pipeline errors, and client-side rendering edge cases.

Can I finish in two weeks? Yes. Option 1 and Option 2 can easily be fully built, styled, and deployed in under two weeks.

Does it show my work the way it needs to be shown? Yes. Static renders display code snippets, pandas dataframe outputs, and architecture diagrams cleanly without performance lags.

4. Final Decision & Written Rationale
Chosen Stack
Option 1: Static HTML5 / CSS3 / Vanilla JS on GitHub Pages

Why I Chose This Stack
I chose pure HTML5, CSS3, and JavaScript on GitHub Pages because my immediate priority is shipping a high-performing, rock-solid portfolio without fighting framework configuration. Since my background is in Software Engineering, writing clean, custom HTML/CSS allows me to have total control over the page layout, rendering speed, and styling without introducing extra build complexity or heavy npm dependencies.

Why I Rejected the Alternatives
Option 2 (Astro/Eleventy): While component templating is nice, setting up SSG build pipelines adds unnecessary friction when my immediate goal is to launch within two weeks.

Option 3 (Next.js/React): Next.js is overkill for a portfolio site that does not yet require client-side state management or dynamic backend rendering. It adds extra JavaScript weight and maintenance overhead for features I do not need right now.

Maintenance & Work Presentation Evaluation
Can I maintain this? Absolutely. Plain HTML/CSS files require zero server maintenance, no dependency updates, and zero risk of breaking build scripts in the future.

Does it show my work well? Yes. It renders my machine learning plots, SQL queries, and API documentation directly, cleanly, and instantaneously on any device.
