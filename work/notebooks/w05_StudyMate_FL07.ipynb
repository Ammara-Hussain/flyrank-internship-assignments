# FINDING THE UPLOADED PDF ON GOOGLE COALAB

```python
from pypdf import PdfReader
import glob

pdf_files = glob.glob("/content/*.pdf")

print("PDF files found:")
for file in pdf_files:
    print(file)

PDF_PATH = pdf_files[0]

reader = PdfReader(PDF_PATH)

notes_text = ""

for page in reader.pages:
    text = page.extract_text()
    if text:
        notes_text += text + "\n"

print("\nStudy material loaded successfully.")
print("File:", PDF_PATH)
print("Number of characters:", len(notes_text))

print("\nPreview:")
print(notes_text[:1000])
```

```python
SYSTEM_INSTRUCTIONS = """
You are StudyMate, a personal AI study coach.

Your job is to help the student study using their uploaded course notes.

CORE WORKFLOW:
1. Understand what topic the student wants to study.
2. Use the uploaded course notes as the primary source.
3. Explain the topic clearly and at an appropriate level.
4. Generate practice questions when requested.
5. Do not reveal quiz answers before the student attempts the questions.
6. Evaluate the student's answers when they provide them.
7. Explain mistakes and identify weak areas.
8. Create a short revision plan based on performance.

GROUNDING:
- The uploaded study material is the primary source.
- Do not invent information and claim that it came from the notes.
- If the requested information cannot be found in the uploaded notes, clearly say:
  "I could not find this information in the uploaded study material."
- Clearly distinguish between information from the notes and additional general knowledge.
- Encourage the student to verify important information against their original course material.

QUIZ BEHAVIOR:
- Generate conceptual and application-based questions.
- Do not show answers before the student attempts the questions.
- Adjust difficulty based on the student's performance.

ANSWER EVALUATION:
- Identify what the student got correct.
- Identify incorrect or incomplete reasoning.
- Explain the correct concept.
- Identify the topic as a weak area when appropriate.

ACADEMIC INTEGRITY:
- If the student says they are currently taking an exam, do not provide direct answers.
- Instead, offer conceptual explanations or study guidance.

STYLE:
- Be supportive and student-friendly.
- Keep explanations clear and organized.
- Use headings and bullet points when useful.
"""

print("StudyMate instructions loaded successfully.")
```

# BUILDING "STUDYMATE" AGENT

```python
def studymate(user_request):
    prompt = f"""
{SYSTEM_INSTRUCTIONS}

UPLOADED COURSE NOTES:
----------------------
{notes_text}
----------------------

STUDENT REQUEST:
{user_request}

Follow the StudyMate instructions carefully.
"""

    response = client.models.generate_content(
        model="gemini-3.6-flash",
        contents=prompt
    )

    return response.text

print("StudyMate agent created successfully.")
```

```python
result = studymate(
    "Explain Dijkstra's algorithm using my uploaded DSA notes. Keep it simple."
)

print(result)
```

# QUIZ FUNCTION

```python
def generate_quiz(topic):
    prompt = f"""
You are StudyMate, a personal study coach.

Use ONLY the uploaded study material to create a quiz about:

{topic}

Create 5 questions:
- 2 conceptual questions
- 2 application-based questions
- 1 slightly challenging question

Do NOT provide the answers.
Do NOT reveal the correct answers in the questions.

Format:

Q1:
Q2:
Q3:
Q4:
Q5:

Remember: The student must attempt the questions before seeing the answers.
"""

    return studymate(prompt)


quiz = generate_quiz("Dijkstra's algorithm")

print("=== STUDYMATE QUIZ ===")
print(quiz)
```

# SIMULATION OF THE STUEDENT'S ANSWERS

```python
student_answers = """
Q1: Dijkstra's algorithm is used to find the shortest path from one source
vertex to other vertices.

Q2: I think negative weights are allowed because the algorithm can simply
calculate the distance again.

Q3: The algorithm chooses the vertex with the smallest current distance.

Q4: It updates the distances of neighboring vertices.

Q5: I am not completely sure about this question.
"""

print("=== STUDENT ANSWERS ===")
print(student_answers)
```

# EVALUATION OF ANSWERS

```python
def evaluate_answers(topic, quiz, answers):
    prompt = f"""
You are StudyMate, a personal study coach.

Topic:
{topic}

Quiz:
{quiz}

Student's answers:
{answers}

Evaluate each answer carefully using the uploaded study material.

For each question:
1. State whether the answer is Correct, Partially Correct, or Incorrect.
2. Explain why.
3. Give the correct concept or reasoning.
4. Identify any misconception.

At the end provide:

WEAK AREAS:
- List the concepts the student needs to revise.

STRONG AREAS:
- List the concepts the student understands.

Do not invent information that is not supported by the uploaded study material.
"""

    return studymate(prompt)


evaluation = evaluate_answers(
    "Dijkstra's algorithm",
    quiz,
    student_answers
)

print("=== STUDYMATE EVALUATION ===")
print(evaluation)
```

# Generate Revision Plan

```python
def create_revision_plan(topic, evaluation):
    prompt = f"""
You are StudyMate, a personal study coach.

Topic:
{topic}

Student evaluation:
{evaluation}

Create a short actionable revision plan based ONLY on the student's
performance.

The plan should contain:

1. Concepts to revise
2. What to practice
3. One recommended self-test
4. Suggested revision priority

Keep the plan concise and practical.
Do not recommend unnecessary topics.
"""

    return studymate(prompt)


revision_plan = create_revision_plan(
    "Dijkstra's algorithm",
    evaluation
)

print("=== STUDYMATE REVISION PLAN ===")
print(revision_plan)
```
